### Analysis 1: How does neighborhood-level socioeconomic status influence Divvy bike accessibility?
*By Jenna Krinsky*

The problem I was focusing on involved creating a more equitable Chicago in terms of Divvy bike accessibility and understanding if there are equal amounts of stations around the city. I first wanted to visualize the distribution of bike stations, as this would show the general concentration of bikes. The data I ended up using, was quite clean as it came from the city of Chicago. There were some irrelevant categories like *Short Name* of stations that I was able to filter out. Additionally, there were a few stations that were outside the city limits so I filtered the stations dataframe based on maximum/minimum longitutude and latitude to produce this geospatial plot of Chicago with Divvy bike stations. 



After that, I wanted to incorporate my main area of analysis: socioeconomic status. So, I merged my existing (above) Chicago dataframe with my demographic dataframe (which included details on race in the Chicago communites *and* income across Chicago) on  "Community" so that each of the physical communities aligned with the demographics of each community. Through this, I wanted to make some initial understandings of the intersection between Divvy stations and incomes in each community. So, I created a choropleths of each income bin (ranging from under $25,000 to making over $125,000) and mapped the stations (white dots) on top of it. 



I quickly discerned that there was a higher density of stations near the wealthier communities, which were lighter on the $125k+ map and darker on the chloropleths of lower income bins. This aligned with a hypothesis that I had: wealthier areas will have more Divvy bike stations, and, thus, greater accessibility to the bikes. My thoughts were two-fold, first that wealthier neighborhoods were in prettier/more-touristy areas so, naturally, there would be more bikes and bike paths there. On top of that, the Divvy corporation, as most corporations do, would most likely prioritize their wealthier customers. Yet, I needed to further explore the station density to show a correlation. Next, I looked at the distribution of income across Chicago communities. 



As expected, there are very large disparities in the city. In some communities, the majority of people make over $125,000 while in others, most people make less than $25,000. Next, I needed to see if these wealthier communities also had more stations. So, I converted and merged the stations onto the map of Chicago to see how many stations were in each community. That distribution is depicted below.



Interestingly, out of the top 10 wealthiest communities *and* the top 10 communities with the most Divvy stations, they share 5 of thes same communities. Those being Lake View, Near North Side, Loop, West Town, and Lincoln Park. This was starting to prove my hypothesis, in that there was a correlation. Of course, I needed to further investigate this. So, I created a heat map to look at the number of stations in each community and how that was connected to the percent of people in each community in each income bracket. 



There is a moderate positive correlation (+0.51) between the percent of people in a community that make over $125,000 and the number of Divvy stations in a community. Yet, for all other income brackets, there is a small, negative correlation. Similarly, for all income brackets, there is a negligible correlation with total population. So although that could be a confouding variable as to why there are more stations in certain areas, it does not have a massive impact. Finally, I zoomed in on the correlation between the wealthiest income bracket and number of stations per community.



Clearly, there is a positive correlation between a higher percentage of people making $125k+ and the number of Divvy stations in the community. With all this in mind, I came to the conclusions that there are very large wealth disparities in Chicago along with a grand variation in the number of Divvy stations in each community. So, I recommend to Divvy stakeholders that they should put more stations in lower socioeconomic areas because this would make the corporation more equitable and increase ridership by bringing more bikes to a diverse subset of people. Overall, Divvy could expand its company by placing around the entire community.

By analyzing this correlation heatmap, we can see the relationship between population density and bike rides is moderately strong (0.49), but the relation between population density and bike stops was totally negligible (0.04). 





Since rides per station tends to increase based on population data -- according to the above scatterplot -- this suggests that the stops are placed to cover geographic area, not be more dense in areas with denser need. Let's explore this further with four density graphs:  



Here we can see that the bike stops mimic the population density, not the density of rides. 



Yet another, even clearer (though the graph is a bit ugly) visualization are these pie charts. The bike rides are not even at all, but bike stations are relatively even. 

#### Potential Conclusion:
This seems to confirm that the Divvy stop placement isn't structured to have more stops in areas with more need, but instead better serve the population overall, mimicking its density. The rides per capita looks similar to the rides per station, meaning that there aren't a strategic placement of bikes -- if there were, the density of rides per station would be pretty even, instead of some being more heavily used and some being used rarely. 

#### Potential Suggestion:
Instead of a general spread, Divvy maybe should aim for strategic placement of stops, aiming to add more in areas where there is more need AND (crucially) in areas where there is POTENTIAL NEED. So where is there potential need? This leeds into my next analysis.
